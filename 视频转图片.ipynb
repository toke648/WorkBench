```python
import cv2
import matplotlib.pyplot as plt
import os

video = cv2.VideoCapture("113761182-1-192.mp4")

num = 0 # 帧数
save_step = 30 # 抽帧间隔
if not os.path.exists("./images"):
    os.mkdir("./images")

while True:
    ret, frame = video.read()
    if not ret:
        break
    num += 1
    if num % save_step == 0:
        cv2.imwrite("./images/" + str(num) + ".jpg", frame)
```
