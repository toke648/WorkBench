```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('data.csv')

plt.rcParams["font.sans-serif"] = ["SimHei"]

print(df)

# 绘制折线图

x  = df['年份']
y1 = df['第一产业']
y2 = df['第二产业']
y3 = df['第三产业']

plt.figure(figsize=(10, 6)) # 设置图形大小和分辨率
plt.plot(x, y1, label="第一产业", marker='o', color='blue') # 绘制折线图
plt.plot(x, y2, label="第二产业", marker='s', color='red')
plt.plot(x, y3, label="第三产业", marker='s', color='green')

plt.legend() # 显示图例
plt.xlabel("年份")
plt.ylabel("生产总值")

plt.show()

```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('data.csv')

plt.rcParams["font.sans-serif"] = ["SimHei"]

print(df)

x  = df['年份']
y1 = df['第一产业']
y2 = df['第二产业']
y3 = df['第三产业']

plt.figure(figsize=(10, 6)) # 设置图形大小和分辨率

plt.subplot(311)
plt.title('产业生产趋势图')
plt.plot(x, y1, label="第一产业", marker='o', color='blue') # 绘制折线图

plt.subplot(312)
plt.title('第二第三产业趋势图')
plt.plot(x, y2, label="第二产业", marker='s', color='red')

plt.subplot(313)
plt.title('第二第三产业趋势图')
plt.plot(x, y3, label="第三产业", marker='s', color='green')

plt.legend() # 显示图例
plt.xlabel("年份")
plt.ylabel("生产总值")

plt.show()

```

```python
# 绘制柱状图

import matplotlib.pyplot as plt
import numpy as np

print(df)

x = ['第一产业', '第二产业', '第三产业']
y = [df['第一产业'][13], df['第二产业'][13], df['第三产业'][13]]

plt.bar(x, y)
plt.title('2013年各产业的收入')
plt.xlabel('产业')
plt.show()

```

```python
# 绘制饼图

import matplotlib.pyplot as plt
import numpy as np

print(df)

# 第一产业最后一个元素
print(df["第一产业"][13])

x = ["第一产业", "第二产业", "第三产业"]
y = [df["第一产业"][13], df["第二产业"][13], df["第三产业"][13]]

plt.figure(figsize=(8, 8))
plt.title("2013年各产业占比")
plt.pie(y, labels=x, autopct="%1.1f%%", shadow=True, startangle=90)
plt.show()

```
