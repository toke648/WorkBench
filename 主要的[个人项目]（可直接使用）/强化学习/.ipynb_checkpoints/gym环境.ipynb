```python
import gym
import os
from matplotlib import pyplot as plt
import time

os.environ["SDL_VIDEODRIVER"] = "dummy" # 这个环境是SDL // KMP_DUPLICATE_LIB_OK ：解决多线程问题

# 创建一个游戏环境
env = gym.make('CartPole-v0')

# 初始化游戏
env.reset()

# 所有可操控对象
print(env.action_space) # 动作空间

print(env.observation_space.high) # 状态空间的上限
print(env.observation_space.low) # 状态空间的下限

print(env.reward_range) # 反馈值空间


```

```python
while True:
    env.step(env.action_space.sample()) # 随机采取一个动作
    print(env.step(env.action_space.sample())) # 随机采取一个动作
    # 显示游戏画面
    display.clear_output(wait=True)
    env.render(mode='rgb_array') # render() ：返回游戏画面 // mode='rgb_array' ：返回游戏画面的数组
    plt.imshow(env.render(mode='rgb_array')) # render() ：返回游戏画面 // mode='rgb_array' ：返回游戏画面的数组
    time.sleep(0.01)
    plt.show() # 显示游戏画面

    # 如果平衡柱倒了，则结束游戏

# 关闭游戏
env.close()
```

```python
import gym
from IPython import display
import time

# 创建月球着陆
env = gym.make('LunarLander-v2')

# 初始化游戏
state, info = env.reset(return_info=True)

# 随机玩N个动作
for i in range(1000):
    action = env.action_space.sample() # 随机选择一个动作
    state, reward, over, info = env.step(action) # 执行动作

    if i % 5 == 0:
        # 显示游戏画面
        display.clear_output(wait=True)
        env.render()
        time.sleep(0.05)

    if over:
        state, info = env.reset(return_info=True)

# 关闭游戏
env.close()
```

```python
import gym
import os
from matplotlib import pyplot as plt

os.environ["SDL_VIDEODRIVER"] = "dummy"  # 避免渲染问题

# 在创建环境时指定 render_mode
env = gym.make('CartPole-v1', render_mode='rgb_array')  # 使用 v1 替代 v0（更稳定）

# 初始化游戏
env.reset()

# 渲染游戏画面
frame = env.render()  # 返回形状为 (1, height, width, channels) 或 (height, width, channels)

# 如果 frame 是四维数组，去掉第一维
if len(frame.shape) == 4:
    frame = frame[0]

plt.imshow(frame)
plt.show()

# 关闭游戏
env.close()
```
