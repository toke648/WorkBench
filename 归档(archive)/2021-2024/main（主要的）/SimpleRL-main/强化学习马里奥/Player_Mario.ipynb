```python
import gym
import pygame
from nes_py.wrappers import JoypadSpace
from gym_super_mario_bros.actions import COMPLEX_MOVEMENT

# 初始化游戏环境
env = gym.make("SuperMarioBros-1-1-v0", apply_api_compatibility=True, render_mode="human")
env = JoypadSpace(env, COMPLEX_MOVEMENT)

# 初始化 Pygame
pygame.init()
screen = pygame.display.set_mode((400, 300))
pygame.display.set_caption("Super Mario Manual Play")

# 按键映射（支持组合动作）
key_mapping = {
    pygame.K_RIGHT: 'right',
    pygame.K_LEFT: 'left',
    pygame.K_a: 'A',  # 跳跃
    pygame.K_b: 'B',  # 冲刺
    pygame.K_UP: 'up',
    pygame.K_DOWN: 'down'
}

# 记录当前按下的按键
pressed_keys = set()

state, _ = env.reset()
done = False

while not done:
    env.render()
    action = []  # 当前帧的动作列表

    # 监听事件
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            done = True
        elif event.type == pygame.KEYDOWN:
            if event.key in key_mapping:
                pressed_keys.add(key_mapping[event.key])  # 记录按下的键
        elif event.type == pygame.KEYUP:
            if event.key in key_mapping:
                pressed_keys.discard(key_mapping[event.key])  # 释放按键

    # 组合动作（例如 ["right", "A"] 表示跳跃+前进）
    action = list(pressed_keys)

    # 找到 `COMPLEX_MOVEMENT` 里对应的索引
    if action in COMPLEX_MOVEMENT:
        action_index = COMPLEX_MOVEMENT.index(action)
    else:
        action_index = 0  # 默认不动

    # 执行动作
    state, reward, done, truncated, info = env.step(action_index)

pygame.quit()
env.close()
```
