```python
import numpy as np
import random

# print(random.random()) # 生成随机数

# print(np.random.randint(10)) # 生成0-9的随机数

# print(np.random.rand(10)) # 生成0-1之间的随机数


probs = np.random.rand(10)

rewards = [[1] for _ in range(10)] # 10个老虎机
# 当抽中时获得奖励，奖励 +1, 不抽中则奖励为0

probs, rewards
```

```python
# 小概率选择一个随机拉杆
def choose_one():
    
    if random.random() < 0.01:
        return random.randint(0, 9)

    # 计算每个老虎机的奖励平均
    rewards_mean = [np.mean(i) for i in rewards]
    print(rewards_mean)

    # 选择期望奖励估值最大的拉杆
    return np.argmax(rewards_mean)

choose_one()
```

```python
def try_and_play():
    i = choose_one()

    rando = random.random()
    print(rando)
    print(probs[i])
    if rando < probs[i]:
        reward = 1
    else:
        reward = 0
    
    rewards[i].append(reward)

    return rewards

try_and_play()
```

```python
def get_result():
    for _ in range(5000):
        try_and_play()
    
    target = probs.max() * 5000

    print(target)
    print(max(probs))

    result = sum(sum(i) for i in rewards)

    return target, result

get_result()
```
